```python
import time
import numpy as np
from multiprocessing import Pool
from concurrent.futures import ProcessPoolExecutor

# --- workload function ---
# Doing fixed-size matrix multiplications to make comparisons fair
def matmul_task(i):
    A = np.random.rand(600, 600)
    B = np.random.rand(600, 600)
    return A @ B

# Number of tasks
N = 50  # Increase for heavier tests / more visible speedup

print("Setup complete. Tasks:", N)
```

```python
np.__config__.show(mode="dicts").keys()
```

```python
start = time.time()

results_seq = [matmul_task(i) for i in range(N)]

end = time.time()
print(f"Sequential execution time: {end - start:.3f} seconds")
```

```python
start = time.time()

with Pool() as p:
    results_pool = p.map(matmul_task, range(N))

end = time.time()
print(f"multiprocessing.Pool time: {end - start:.3f} seconds")
```

```python
start = time.time()

with ProcessPoolExecutor() as ex:
    results_exec = list(ex.map(matmul_task, range(N)))

end = time.time()
print(f"ProcessPoolExecutor time: {end - start:.3f} seconds")
```
