```python
def terminal(width, length, depth) -> list[float]:
    """From the width and length return the atom indices highligting the terminal atoms of depth `d`"""
    atom_idx = []
    for i in range(1, width+1):
        for d in range(depth):
            atom_idx.append(2*i*length - (1+d))
    return atom_idx

WIDTH = 10
LENGTH = 15
DEPTH = 4
AXES = "xy"

terminal_atom_idx = terminal(WIDTH, LENGTH, DEPTH)
full_ribbon = make_nanoribbon(width=WIDTH, length=LENGTH)
full_ribbon.plot(axes=AXES, atoms_style={"atoms": terminal_atom_idx, "color": "red"})
```
